# Step1 : 入力列から状態軌道と総コストを求める

初期状態と入力列から、離散時間状態方程式を順方向に計算すると状態軌道が得られ、その軌道から総コストが決まる。このことを説明する。

## 離散時間状態方程式

連続時間のシステムを以下のように定義する。$x,u$は多変数を想定している。

$$
\dot{x}(t) = f_c(x(t), u(t))
$$

刻み時間 $\Delta t$ で離散化すると、以下となる。

$$
x_{k+1} = f(x_k, u_k)
$$

この $f$ が離散時間の状態遷移を表す関数である。



例えば連続モデルを以下とする。

$$
m \ddot{p} = u - k p - k_3 p^3
$$

位置$p$, 速度$v$ からなる状態ベクトルを $x=[p , v]^T$ とすると、連続時間状態方程式は以下となる。

$$
\dot{x} = \begin{bmatrix} \dot{p} \\ \dot{v} \end{bmatrix}
= f_c(x, u) = 
\begin{bmatrix}
v \\ \frac{1}{m}\left( u - k p - k_3 p^3 \right)
\end{bmatrix}
$$

これを前進Euler法で離散化すると次となる。

$$
\begin{aligned}
p_{k+1} &= p_k + \Delta t v_k \\
v_{k+1} &= v_{k} + \frac{\Delta t}{m}\left( u_k - k p_k - k_3 {p_k}^3 \right)
\end{aligned}
$$

よって離散時間状態方程式は以下のように現在の状態が $f$ に含まれた形になる。

$$
x_{k+1} = f(x_k, u_k) = \begin{bmatrix}
p_k + \Delta t v_k \\
v_{k} + \frac{\Delta t}{m}\left( u_k - k p_k - k_3 {p_k}^3 \right)
\end{bmatrix}
$$

## 入力列と状態軌道

予測ホライゾンを$N$ステップに離散化すると、入力列$U$は以下となる。

$$
U = [u_0, u_1, \cdots, u_{N-1}]
$$

この $U$ により状態を順番に計算する。

$$
x_{k+1} = f(x_k, u_k) , \quad k = 0, \cdots , N-1
$$

得られる状態軌道$X$は、$N+1$個になる。

$$
X = [x_0, x_1, \cdots, x_N]
$$

## 総コスト

総コスト$J$は離散状態で次のようになる。
連続時間の積分コストを離散化しているため、ステージコストには $\Delta t$ を掛ける。

$$
J = \phi(x_N) + \sum_{k=0}^{N-1} \Delta t \ \ell(x_k, u_k)
$$

- $\ell (x_k, u_k)$ : 途中の各時刻で発生するステージコスト
- $\phi(x_N)$ : 最終の状態に対する終端コスト


ステージコスト$\ell$と終端コスト$\phi$ は、例えば次のような2次形式で定義できる。ここで $x$は状態ベクトル, $u$は入力ベクトルであり、$Q,Q_f,R$はそれぞれの重み行列である。

$$
\begin{aligned}
\ell(x_k, u_k) &= \frac{1}{2}\left(x_k - x_{ref} \right)^T \ Q \ \left(x_k - x_{ref} \right) + \frac{1}{2}{u_k}^T R \ u_k \\
\phi(x_N) &= \frac{1}{2} \left(x_N - x_{ref} \right)^T \ Q_f \ \left(x_N - x_{ref} \right)
\end{aligned}
$$


状態軌道$X$ は以下のように初期状態 $x_0$ と入力列 $U$ の関数である。

$$
X = X(U; x_0)
$$

初期状態 $x_0$ を決めれば、入力列 $U$ を決めるだけで状態軌道$X$が決まる。したがって、総コスト $J$ も$U$の関数として表せる。

$$
J = J(U; x_0)
$$

iLQRはこの入力列$U$を改善していく。
